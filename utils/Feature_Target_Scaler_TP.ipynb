```python
import pandas as pd
import numpy as np
import joblib
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('KC_6H_2001-2025_Final.csv')
df.info()
```

```python
# Getting the index of the first occurrence of each year and printing each year and its corresponding index
first_indices = df.groupby('year').head(1).index
first_indices_list = first_indices.tolist()
for year, index in zip(df['year'].unique(), first_indices_list):
    print(f"Year: {year}, First Index: {index}")
```

```python
# Setting validation data from 2019 to 2020, and test data from 2021 to 2025
val_idx = 10097664
test_idx = 11220480
```

```python
target_variable = 'tp'
cols_to_drop = ['latitude', 'longitude', 'year']
X_df = df.drop(columns=cols_to_drop)
y_df = df[[target_variable]]
```

```python
y_np_fit = np.log1p(y_df.values[:val_idx])
X_np_fit = X_df.values[:val_idx].copy()
X_np_fit[:, X_df.columns.get_loc('tp')] = np.log1p(X_np_fit[:, X_df.columns.get_loc('tp')])
if 'ws' in X_df.columns:
    X_np_fit[:, X_df.columns.get_loc('ws')] = np.log1p(X_np_fit[:, X_df.columns.get_loc('ws')])

feature_scaler = MinMaxScaler().fit(X_np_fit)
target_scaler = MinMaxScaler().fit(y_np_fit)
```

```python
joblib.dump(feature_scaler, 'feature_scaler_tp.pkl');
joblib.dump(target_scaler, 'target_scaler_tp.pkl');
```
